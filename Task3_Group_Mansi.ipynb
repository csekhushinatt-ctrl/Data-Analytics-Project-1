```python
drink_name = input("Enter drink name: ")
drink_price = float(input("Enter drink price: "))
snack_price = float(input("Enter snack price: "))
drink_qty = int(input("Enter drink quantity: "))
snacks_qty = int(input("Enter snacks Quantity: "))


drink_name = drink_name.strip()
drink_name = drink_name.upper()

drink_cost = drink_price * drink_qty
snack_cost = snack_price * snacks_qty
subtotal = drink_cost + snack_cost
service_fee = subtotal * 0.10
final_total = subtotal + service_fee

budget = float(input("Enter customer budget: "))
balance = budget - final_total

final_total = round(final_total, 2)
balance = round(balance, 2)

can_afford = budget >= final_total
earns_reward = final_total >= 15

print("Drink Name is: ", drink_name)
print("Drink Price is: ", drink_price)
print("Snack Price is: ", snack_price)
print("Drink Quantity is: ",drink_qty)
print("Snack Quantity is: ",snacks_qty)
print("Total Drink Cost:", drink_cost)
print("Total Snack Cost:", snack_cost)
print("Subtotal:", subtotal)
print("Service Fee:", service_fee)
print("Final Total:", final_total)
print("Remaining Balance:", balance)
print("Can Afford:", can_afford)
print("Earns Reward:", earns_reward)
print(f"Final Total: {final_total}")
print(f"Remaining Balance: {balance}")
```
