```python
book_title = "python for beginners"
pages_text = "180"
rating_text = "4.5"
loan_status_text = "False"
x = book_title.upper()
y = book_title.lower()
pages = int(pages_text)
rating = float(rating_text)
loan_status = bool(loan_status_text)
add_new_pages = pages + 20
new_text_pages = str(add_new_pages)

print("Book Title:",book_title)
print("Page Text:",pages_text)
print("Rating Text:",rating_text)
print("Loan Status Text:",loan_status_text)
print("Book Title:",x)
print("Book Title:",y)
print(type(pages))
print(type(rating))
print(type(loan_status))
print("Add New Pages:",add_new_pages, "Type:", type(new_text_pages))
print("bool(0):",bool(0))
print("bool(1):",bool(1))
print("bool(""):",bool(""))
print("bool(loan_status_text):" , bool(loan_status))

# F-bool("False") returns True because "False" is a nonempty string.
#  Python checks whether the string is empty, rather than interpreting
#   the word inside it. An empty string, "", returns False.
```
