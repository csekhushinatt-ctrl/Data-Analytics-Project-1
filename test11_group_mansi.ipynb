```python
travel_mode = input("Enter your travel Mode(Bike,Walk,or Bus):")
One_way_distance = float(input("Enter One-way distance in km:"))
Number_of_commuting_days = int(input("Enter number of commuting days:"))

travel_mode = travel_mode.strip().lower()

daily_round_trip_distance = One_way_distance * 2
weekly_distance = daily_round_trip_distance * Number_of_commuting_days
bus_fare = weekly_distance * 0.30
# total_distance = daily_round_trip_distance * weekly_distance

active_mode = (travel_mode == "walk" or travel_mode == "bike")

active_goal_met =active_mode and (weekly_distance>=15)
goal_not_met = not active_goal_met

print(f"Travel Mode:{travel_mode}")
print(f"Daily Round Trip distance:{daily_round_trip_distance}")
print(f"Weekly Distance:{weekly_distance}")
# print(f"Total distance:{total_distance}")
print(f"Bus Fare:{bus_fare}")
print(f"Active Mode:{active_mode}")
print(f"Active Goal Met:{active_goal_met}")
print(f"Goal Not Met:{goal_not_met}")
```
