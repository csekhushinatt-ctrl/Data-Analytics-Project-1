```python
name = input("Enter your first and last name: ")
age = int(input("Enter your age: "))
course = input("Enter your favorite course: ")

# Split the name before using its parts
namepart = name.strip().split()
firstname = namepart[0].upper()
lastname = namepart[1].upper()

full_name = firstname + " " + lastname
next_year_age = int(age) + 1
lenght_of_name = len(full_name)

print("Firstname:", firstname)
print("Lastname:", lastname)
print("Full name:", full_name)
print("Age:", age)
print("Course:", course)
print("Next year age:", next_year_age)
print("Length of name:", lenght_of_name)
print(f"My name is {full_name}.")
print(f"I am {age} years old.")
print(f"My favorite course is {course}.")

print(type(age))
print(type(full_name))
```
